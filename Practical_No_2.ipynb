# Theory

Artificial Neural Networks (ANNs) are one of the fundamental deep learning models used to learn complex relationships between input features and target variables. An ANN consists of an input layer, one or more hidden layers, and an output layer. During training, the network learns patterns from the data by adjusting its weights through backpropagation and optimization algorithms such as Adam.

Before training an ANN, the dataset must be preprocessed by encoding categorical variables into numerical values and normalizing numerical features to improve convergence and training performance. Proper preprocessing ensures that the model learns efficiently and produces reliable predictions.

A **baseline ANN model** serves as the initial reference model for evaluating future improvements. Rather than immediately building complex architectures, a simple ANN is trained first to establish a performance benchmark.

**Experiment tracking** is an important practice in deep learning projects. It involves recording hyperparameters such as learning rate, batch size, and epochs along with the corresponding evaluation metrics. Tracking multiple experiments enables systematic comparison of different model configurations, improves reproducibility, and helps identify the best-performing model.

In this experiment, the Crop Yield Prediction dataset is used to build a baseline ANN model for predicting crop yield. Multiple experiments are conducted using different hyperparameter settings, and their results are stored in an experiment log for comparison. Finally, the best-performing model is saved for future prediction and deployment.


**Step 1: Import Required Libraries**

```python
# Data Manipulation
import pandas as pd
import numpy as np

# Data Preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Deep Learning
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# Evaluation Metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Experiment Tracking
import datetime
```

**Step 2: Load the Crop Yield Dataset**

```python
# ==========================================
# Step 2.1: Upload Dataset from Local Computer
# ==========================================

from google.colab import files

uploaded = files.upload()
```

```python
# ==========================================
# Step 2.2: Load the Dataset
# ==========================================

import pandas as pd

# Get the uploaded file name
file_name = list(uploaded.keys())[0]

# Load the dataset
df = pd.read_csv(file_name)

# Display the first 5 rows
df.head()
```

**Step 3: Explore the Dataset**

```python
# Display the first 5 rows
print("First 5 Rows:")
display(df.head())

# Display dataset shape
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display dataset information
print("\nDataset Information:")
df.info()

# Display summary statistics
print("\nSummary Statistics:")
display(df.describe())
```

**Step 4: Select Features (X) and Target (y)**

```python
# Input Features
X = df.drop("Yield_tons_per_hectare", axis=1)

# Target Variable
y = df["Yield_tons_per_hectare"]

# Display shapes
print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)

# Display first 5 rows of features
display(X.head())

# Display first 5 target values
display(y.head())
```

**Step 5: Encode Categorical Features**

```python
# Convert categorical variables into numerical variables
X = pd.get_dummies(X, drop_first=True)

# Display the new shape
print("Encoded Features Shape:", X.shape)

# Display the first 5 rows
display(X.head())
```

**Step 6: Split the Dataset into Training and Testing Sets**

```python
# Split dataset into Training (80%) and Testing (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shape of each dataset
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)
```

**Step 7: Normalize (Scale) the Features**

```python
# Create StandardScaler object
scaler = StandardScaler()

# Fit and transform the training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform the testing data
X_test_scaled = scaler.transform(X_test)

# Display the shape of scaled datasets
print("Scaled Training Features:", X_train_scaled.shape)
print("Scaled Testing Features :", X_test_scaled.shape)
```

**Step 8: Build the Baseline ANN Model**

```python
# Create the ANN model
model = Sequential([

    # Input Layer + First Hidden Layer
    Dense(128, activation='relu', input_shape=(X_train_scaled.shape[1],)),

    # Dropout Layer
    Dropout(0.2),

    # Second Hidden Layer
    Dense(64, activation='relu'),

    # Output Layer
    Dense(1)
])

# Display model summary
model.summary()
```

**Step 9: Compile the ANN Model**

```python
# Compile the model
model.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mae']
)

print("ANN Model Compiled Successfully!")
```

**Step 10: Train the Baseline ANN Model**

```python
# Train the ANN model
history = model.fit(
    X_train_scaled,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

**Step 11: Evaluate the Model**

```python
# Evaluate the model on the test dataset
test_loss, test_mae = model.evaluate(X_test_scaled, y_test, verbose=1)

print("Test Loss (MSE):", round(test_loss, 4))
print("Test MAE:", round(test_mae, 4))
```

**Step 12: Manual Experiment Tracking**

```python
# Create an empty list to store experiment details
experiment_log = []

# Store the first experiment
experiment_log.append({
    "Experiment_ID": "EXP-01",
    "Learning_Rate": 0.001,
    "Epochs": 20,
    "Batch_Size": 32,
    "Test_Loss_MSE": round(test_loss, 4),
    "Test_MAE": round(test_mae, 4)
})

# Display the experiment log
log_df = pd.DataFrame(experiment_log)
display(log_df)
```

**Step 13: Run Multiple Hyperparameter Configurations**

```python
# Create a new ANN model
model_exp2 = Sequential([
    Dense(128, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dense(1)
])

# Compile the model with a different learning rate
model_exp2.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0005),
    loss='mean_squared_error',
    metrics=['mae']
)

# Train the model
history2 = model_exp2.fit(
    X_train_scaled,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# Evaluate the model
test_loss2, test_mae2 = model_exp2.evaluate(
    X_test_scaled,
    y_test,
    verbose=0
)

# Save experiment details
experiment_log.append({
    "Experiment_ID": "EXP-02",
    "Learning_Rate": 0.0005,
    "Epochs": 5,
    "Batch_Size": 32,
    "Test_Loss_MSE": round(test_loss2,4),
    "Test_MAE": round(test_mae2,4)
})

print("EXP-02 Completed Successfully!")
```

**Step 14: Save Experiment Log (experiment_log.csv)**

```python
# Convert experiment log into a DataFrame
log_df = pd.DataFrame(experiment_log)

# Display the experiment log
display(log_df)

# Save the experiment log
log_df.to_csv("experiment_log.csv", index=False)

print("Experiment log saved successfully as 'experiment_log.csv'")
```

**Step 15: Compare the Results**

```python
print("Experiment Comparison:")
display(log_df)

# Find the best experiment based on lowest Test MAE
best_experiment = log_df.loc[log_df["Test_MAE"].idxmin()]

print("\nBest Experiment:")
display(best_experiment)
```

**Step 16: Save the Best Model (baseline_ann.keras)**

```python
# Save the best model
model_exp2.save("baseline_ann.keras")

print("Best ANN Model Saved Successfully!")
```
